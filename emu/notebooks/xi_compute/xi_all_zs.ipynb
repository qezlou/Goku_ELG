## Vizsualaize the $\xi(r)$ for all redshifst:

```python
import h5py
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8,6))


fid = 'L2'
box = {'L2': 250, 'HF':1000}
parts = {'L2': 750, 'HF':3000}
sim_id = 123

data_dir = '/home/qezlou/HD2/HETDEX/cosmo/data/xi_on_grid'

for z in [0.0, 0.2, 0.5, 1.0, 2.0, 2.5, 3.0, 4.0, 9.0]:
    fname = f'all_compressed_10p_Box{box[fid]}_Part{parts[fid]}_z{z}.hdf5'

    with h5py.File(f'{data_dir}/{fname}', 'r') as f:
        ind = list(f['sim_id'][:]).index(sim_id)
        print(f['sim_tags'][ind])
        ax.plot(f['mbins'][:], f['corrs'][ind,-1,:], label=f'z = {z}')
ax.legend()
ax.set_xscale('log')
ax.set_yscale('log')
```
